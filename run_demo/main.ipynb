```python
import duckdb
```

```python
conn: duckdb.DuckDBPyConnection = duckdb.connect()
conn.execute("INSTALL ducklake;")
conn.execute("SET s3_endpoint='idmlakehouse.tmslab.cn';")
conn.execute("SET s3_url_style='path';")
conn.execute("SET s3_use_ssl='false';")
conn.execute("""
ATTACH 'ducklake:./metadata.ducklake' as hea (
    DATA_PATH 's3://idmdatabase/hea'
);
""")
conn.execute("use hea;")
```

```python
# Show all table names in the database
conn.sql("SHOW TABLES;")
```

```python
# Show all combinations of elements
conn.sql("SELECT * FROM hea_elements_6")
```

```python
# Show all combinations of composition proportions
conn.sql("SELECT * FROM hea_con_6;")
```

```python
# Show all descriptor names
conn.sql("SELECT * FROM descriptor_names;")
```

```python
# Show some columns of a table
conn.sql("SELECT con_index, ave_fe1, rmse_ft2, range_fp5 FROM hea_6_c_120;")
```
